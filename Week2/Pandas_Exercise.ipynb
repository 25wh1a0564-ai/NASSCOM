```python
import pandas as pd
import numpy as np

print("\n--- Task 2: Synthetic Data Generation for Practice ---")

# Set seed for reproducible results
np.random.seed(1)

# Create 100 dates
dates = pd.date_range(start='2023-01-01', end='2023-04-09', periods=100)

# Create DataFrame
df_sales = pd.DataFrame({
    'Date': dates,
    'Region': np.random.choice(['East', 'West', 'North', 'South'], size=100),
    'Product': np.random.choice(['Laptop', 'Mouse', 'Keyboard', 'Monitor'], size=100),
    'Sales': np.random.randint(100, 1001, size=100),
    'Quantity': np.random.randint(1, 11, size=100)
})

# Introduce NaN values
sales_nan = np.random.choice(df_sales.index, size=5, replace=False)
quantity_nan = np.random.choice(df_sales.index, size=3, replace=False)

df_sales.loc[sales_nan, 'Sales'] = np.nan
df_sales.loc[quantity_nan, 'Quantity'] = np.nan

# Display the DataFrame
print("\nSales DataFrame:")
print(df_sales)
```

```python
# Task 3: Indexing and Slicing in DataFrames
print("\n--- Task 3: Indexing and Slicing in DataFrames ---")

# 3.1 Select Region and Sales columns
df3_1 = df_sales[['Region', 'Sales']]
print("\n3.1 Region and Sales:")
print(df3_1)

# 3.2 Select rows where Product is Laptop
df3_2 = df_sales[df_sales['Product'] == 'Laptop']
print("\n3.2 Product = Laptop:")
print(df3_2)

# 3.3 Select rows where Region is East AND Sales > 500
df3_3 = df_sales[
    (df_sales['Region'] == 'East') &
    (df_sales['Sales'] > 500)
]
print("\n3.3 East and Sales > 500:")
print(df3_3)

# 3.4 Select Sales value for the first Monitor
monitor_rows = df_sales[df_sales['Product'] == 'Monitor']

# Using .loc
first_monitor_index = monitor_rows.index[0]
val3_4 = df_sales.loc[first_monitor_index, 'Sales']

# Using .iloc
val3_4_iloc = monitor_rows.iloc[0]['Sales']

print("\n3.4 First Monitor Sales using .loc:", val3_4)
print("First Monitor Sales using .iloc:", val3_4_iloc)
```

```python
# Task 4: Data Cleaning (Handling Missing Values)
print("\n--- Task 4: Data Cleaning (Handling Missing Values) ---")

# 4.1 Check missing values in each column
missing_counts4_1 = df_sales.isnull().sum()

print("\n4.1 Missing values:")
print(missing_counts4_1)


# 4.2 Fill missing Sales values with the mean
df4_2_filled = df_sales.copy()

sales_mean = df4_2_filled['Sales'].mean()
df4_2_filled['Sales'] = df4_2_filled['Sales'].fillna(sales_mean)

print("\n4.2 DataFrame after filling missing Sales:")
print(df4_2_filled)


# 4.3 Drop rows with any missing values
df4_3_dropped = df_sales.dropna()

print("\n4.3 DataFrame after dropping rows with missing values:")
print(df4_3_dropped)
```

```python
# Task 5: Data Manipulation (Columns, Sorting, Grouping)
print("\n--- Task 5: Data Manipulation (Columns, Sorting, Grouping) ---")

# 5.1 Create Total_Revenue column
df_sales['Total_Revenue'] = df_sales['Sales'] * df_sales['Quantity']

print("\n5.1 Total Revenue:")
print(df_sales)


# 5.2 Sort by Date ascending, then Sales descending
df5_2_sorted = df_sales.sort_values(
    by=['Date', 'Sales'],
    ascending=[True, False]
)

print("\n5.2 Sorted DataFrame:")
print(df5_2_sorted)


# 5.3 Group by Region and calculate sum of Sales and Quantity
df5_3_grouped_region = df_sales.groupby('Region')[['Sales', 'Quantity']].sum()

print("\n5.3 Grouped by Region:")
print(df5_3_grouped_region)


# 5.4 Group by Product and calculate average Sales
df5_4_grouped_product = df_sales.groupby('Product')['Sales'].mean()

print("\n5.4 Average Sales by Product:")
print(df5_4_grouped_product)
```
